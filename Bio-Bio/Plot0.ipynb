```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import kaleido
import matplotlib.pyplot as plt


data_folder = "/home/j/Documents/IIASA/MM-Emulator"
data_file = "magpie_input_MP00BI00nov23.csv"
figure_folder = "/".join([data_folder, "figures/"])
data_file_path = "/".join([data_folder,data_file])
data = pd.read_csv(data_file_path)
```

```python
df = data[(data['Region']=='World') & (data['GHGscen']=="GHG000") & (data['Variable']=="Primary Energy|Biomass")]
df = df.drop(columns=["Region", "SSPscen", "GHGscen", "SDGscen", "1995", "2000", "2005", "2010", "2015"])
print(df)
```

```python
bv = df['2100'].values
bv_rev = bv * -1
print(bv_rev)
```

```python
ghg = data[(data['Region']=='World') & (data['BIOscen']=="BIO00") & (data['Variable']=="Price|Carbon|CO2")]
ghg = ghg.drop(columns=["Region", "SSPscen", "BIOscen", "SDGscen", "1995", "2000", "2005", "2010", "2015"])
gv = ghg['2025'].values
print(gv)
```

```python
co2 = data[(data['Region']=='World') & (data['Variable']=='Emissions|CO2|AFOLU')]
co2['cumulative'] = co2['2025'] * 5 + co2['2030'] * 5 + co2['2035'] * 5 + co2['2040'] * 5 + co2['2045'] * 5 + co2['2050'] * 5 + co2['2055'] * 5 + co2['2060'] * 5 + co2['2070'] * 10 + co2['2080'] * 10 + co2['2090'] * 10 + co2['2100'] * 10
print(co2['cumulative'])
```

```python
ghgscens = co2[co2['BIOscen']=="BIO00"]['GHGscen'].values
print(ghgscens)
emis = []
#ghgscens <- dplyr::filter(co2,BIOscen=="BIO00")$GHGscen
for g in ghgscens:
   # print(co2[co2['GHGscen']==g]['cumulative'].values/1000)
    emis.append(co2[co2['GHGscen']==g]['cumulative'].values/1000)
emis_df = pd.DataFrame(emis)
emis_df
#print(emis_df)
```

```python
#print(emis_df.values)
```

```python
fig = go.Figure(
    layout = go.Layout(
        margin=dict(
            l=0,
            r=0,
            b=0,
            t=0
        ),
        scene=dict(
            xaxis=dict(
                title = '',
                autorange="reversed",
                gridcolor='White'
            ),
            yaxis=dict(
                title=""
            ),
            zaxis=dict(
                title=""
            ),
            camera = dict(
                center = dict(
                    z = -0.2,
                    x = 0.05
                ),
                eye = dict(
                    x = 2.6,
                    y = 1.4,
                    z = 1.4                
                )
            )            
        )
    )
)

fig.add_trace(
    go.Surface(
        x = bv,
        y = gv,
        z = emis_df.values,        
        contours=dict(
            x = dict(
                show = True,
                color = "White"
            ),
            y = dict(
                show = True,
                color = "White"
            )
        )        
    )
)

fig.update_layout(
    template='plotly_white',
    plot_bgcolor = "Green",
    scene = dict(
        bgcolor = "White"
    )
)



fig.show()
#print(fig)
fig.write_image("/home/j/Documents/IIASA/MM-Emulator/figures/fig0_surface.png", width=1200, height = 1000)
```
